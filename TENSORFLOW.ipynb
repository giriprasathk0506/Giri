```python
import tensorflow as tf

x = tf.Variable(0, name="x")
y = tf.Variable(10, name="y")
print("Initial Values")
print("x =", x.numpy())
print("y =", y.numpy())
for i in range(5):
    x.assign_add(1)
    y.assign_add(2)
print("\nUpdated Values")
print("x =", x.numpy())
print("y =", y.numpy())
```

Save Variables into a Checkpoint

```python
checkpoint = tf.train.Checkpoint(x=x, y=y)
save_path = checkpoint.save('./checkpoints/my_checkpoint')
print("Checkpoint saved at:")
print(save_path)
```

Clear Memory (Simulate Notebook Restart)

```python
del x
del y
print("Variables cleared from memory.")
```

Reload Variables from Checkpoint

```python
x = tf.Variable(0, name="x")
y = tf.Variable(0, name="y")
checkpoint = tf.train.Checkpoint(x=x, y=y)
checkpoint.restore(tf.train.latest_checkpoint('./checkpoints'))
print("Restored Values")
print("x =", x.numpy())
print("y =", y.numpy())
```
