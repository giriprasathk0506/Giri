```python
import pandas as pd
df=pd.read_csv("/content/sample_data/IPL.csv")
df
```

SORTING

```python
df.columns=df.columns.str.strip()
df.sort_values(by="Player",ascending=True)
```

```python
df.sort_values(by="Strike_Rate")
```

```python
df.sort_values(by=["Player","Runs"])
```

CREATING NEW COLUMN


```python
df["Tournament"] = "IPL"
df
```

```python
df['WINNER_LENGTH'] = df['Player'].str.len()
df
```

GROUP BY

```python
team_runs = df.groupby("Team")["Runs"].sum()
team_runs
```

```python
avg_runs = df.groupby("Team")["Runs"].mean()
avg_runs
```

```python
result = df.groupby("Team")["Runs"].agg(["sum", "mean", "max", "min", "count"])
result
```

FILTERING RECORDS BASED ON CONDITIONS

```python
result = df[df['Runs'] > 50]
result
```

```python
result = df[(df['Team'] == 'MI') | (df['Team'] == 'CSK')]
result
```
